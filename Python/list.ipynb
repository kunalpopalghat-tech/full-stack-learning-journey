## list data type :-
### Q. what is list in python ?
##### List is mutable heterogenius collection of elements where order is maintaned and duplicates are allowed
##### - List values inclosed in square bracket [] .
##### mutable-------> changable, growable
##### heterogenius collection of elements------->int, float, bool, string, complex, list, tuple, dict.
##### order is maintained-------> index is present (indexing & slicing)
##### Duplicates are allowed

## Q. how to access data from list ?
    by using indexing and slicing we can access data from list

```python
# using indexing
print(batch_1341_db[0]) # it shows first element of list whose index is 0.
print(batch_1341_db[-1]) # it shows last element of list whose index is -1 or lenght - 1.
```

```python
# using slicing
print(batch_1341_db[0])
print(batch_1341_db[2 : 5])
```

```python
# print list in reverse order using slicing
print(batch_1341_db[::-1])
```

## Q. How to iterate every element from list ?
    by using loops -----> for, while

```python
filtered_list = []
for name in batch_1341_db:
    if isinstance(name,str):
        filtered_list.append(name)

print(filtered_list)
```
