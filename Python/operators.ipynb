# 1.Arithmetic operator : 
### Arithmetic operators are used to perform mathematical operation on numeric values.
### + , - , * , / , % , // , **.

#### 1.Addition(+) :
##### Addition is used to add to values.

```python
#Example :
a=10
b=5
print(a+b)
```

#### 2.Substraction(-) :
##### Used to substract one value from another value.

```python
#Example :
a=20
b=10
print(a-b)
```

#### 4.Multiplication(*) :
##### It used to multiply values.

```python
#Example :
a=10
b=5
print(a*b)
```

#### 5.Division(/) :
##### Used to divide one value by another.

```python
#Example :
a=10
b=2
print(a/b) # Division always return a floating values.
```

#### 6.Modulus(%):
##### The modulas operator returns the reminder after division.

```python
#Example 
print(10%3)
# Because :
#quotient=3
#reminder=1
```

#### 7.Floor division(//):
##### Floor division returns the floor value of the division result (It return only integer value of the division result)

```python
#Example :
print(10//3)
```
