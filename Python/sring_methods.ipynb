## String Methods

String methods are built-in methods used to perform different operations on strings.

Syntax:
string.method()

```python
#example
name = "python"

print(name.upper()) # upper  is method which converts sring into uppercase
```

## methods :-
#### 1. upper()

The upper() method converts all lowercase characters in a string into uppercase characters.

Syntax:
string.upper()

```python
# example
name = "python programming"

print(name.upper())
```

#### 2. lower()

The lower() method converts all uppercase characters in a string into lowercase characters.

Syntax:
string.lower()

```python
# example
name = "PYTHON PROGRAMMING"

print(name.lower())
```

#### 3. capitalize()

The capitalize() method converts the first character of a string to uppercase and converts the remaining characters to lowercase.

Syntax:
string.capitalize()

```python
# example
text = "pYTHON programming"

print(text.capitalize())
```

#### 4. title()

The title() method converts the first character of each word to uppercase.

Syntax:
string.title()

```python
# example
text = "python programming language"

print(text.title())
```

### Difference between capitalize and title :
##### capitalize() → First character of the string
##### title()      → First character of every word

#### 5. strip()

The strip() method removes whitespace from the beginning and end of a string.

Syntax:
string.strip()

```python
# example
name = "   Kunal   "

print(name.strip())
```

#### 6. lstrip()

The lstrip() method removes whitespace from the beginning (left side) of a string.

```python
# example
name = "   Kunal"

print(name.lstrip())
```

#### 7. rstrip()

The rstrip() method removes whitespace from the end (right side) of a string.

```python
# example
name = "Kunal   "

print(name.rstrip())
```

#### 8. replace()

The replace() method replaces a specified character or substring with another character or substring.

Syntax:
string.replace(old, new)

```python
text = "I love Java"

result = text.replace("Java", "Python")

print(result) #replace() original string बदलत नाही. तो new string return करतो.
```

#### 9. count()

The count() method returns the number of times a specified character or substring occurs in a string.

Syntax:
string.count(value)

```python
# example 1
text = "banana"

print(text.count("a")) # it shows count of "a" character.
```

```python
# example 2
text = "Python Python Python"

print(text.count("Python")) # it shows count of "python" character.
```

#### 10. find()

The find() method returns the index position of the first occurrence of a specified substring.

Syntax:
string.find(value)

```python
# example 
text = "Python"

print(text.find("t")) # it returns index number of "t" character
```

```python
# example 2
text = "Python"

print(text.find("z")) # if find value is not found it returns -1 .
```

#### 11. index()

The index() method returns the index position of the first occurrence of a specified substring.

Syntax:
string.index(value)

```python
# example
text = "Python"

print(text.index("t")) # it is similar like find() . it is also return index number of given character
```

### Important Difference between find() and index().
 in find() if value is not found it gives output -1 and in index() if value is not found it shows error.

```python
# example of find()
text = "Python"

print(text.find("z")) # if value is not found it gives output -1 .
```

```python
# example of index()
text = "Python"

print(text.index("z")) # if value is not found it shows error .
```

#### 12. split()

The split() method splits a string into a list of substrings based on a specified separator.

Syntax:
string.split(separator)

```python
# example
text = "I love Python"

result = text.split()

print(result) 
```

```python
# example 1
text = "apple,banana,mango"

print(text.split(","))
```

#### 13. join()

The join() method joins the elements of an iterable into a single string using a specified separator.

Syntax:
separator.join(iterable)

```python
# example
words = ["I", "love", "Python"]

result = " ".join(words)

print(result)
```

```python
# example 1
words = ["Apple", "Banana", "Mango"]

result = ", ".join(words)

print(result)
```

### replace() → Replace
### count()   → Count
### find()    → Find position
### index()   → Find position
### split()   → String → List
### join()    → List → String

#### 14. startswith()

The startswith() method checks whether a string starts with a specified value.

It returns True if the string starts with the specified value, otherwise it returns False.

Syntax:
string.startswith(value)

```python
# example
text = "Python Programming"

print(text.startswith("Python")) 
```

```python
# example 
text = "Python Programming"

print(text.startswith("Java"))
```

#### 15. endswith()

The endswith() method checks whether a string ends with a specified value.

It returns True if the string ends with the specified value, otherwise it returns False.

Syntax:
string.endswith(value)

```python
# example
text = "Python Programming"

print(text.endswith("Programming"))
```

```python
# example
text = "Python Programming"

print(text.endswith("Python"))
```

#### 16. isalpha()

The isalpha() method returns True if all characters in a string are alphabetic characters.

It returns False if the string contains spaces, numbers, or special characters.

```python
# examplt
text = "Python"

print(text.isalpha())
```

```python
# example
text = "Python123"

print(text.isalpha())
```

#### 17. isdigit()

The isdigit() method returns True if all characters in a string are digits.

Otherwise, it returns False.

```python
# example
number = "12345"

print(number.isdigit())
```

```python
# example
number = "123abc"

print(number.isdigit())
```

#### 18. isalnum()

The isalnum() method returns True if all characters in a string are either alphabetic characters or digits.

It returns False if the string contains spaces or special characters.

```python
# example
text = "Python123"

print(text.isalnum())
```

```python
# example
text = "Python 123" # it gives false because space is include in the string

print(text.isalnum()) 
```
