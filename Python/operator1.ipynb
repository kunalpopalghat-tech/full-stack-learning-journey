# Operators
## Types of operators:
### 2.Relational / Comparison operator
#### > , < , >= , <= .
#### Equality operators
#### ==(Equal to equal) , !=(Not equal to) .
#### Relational operator always returns a boolean values(True/False)

```python
#Example
sagar=21
omkar=19
print(omkar > sagar) #False because omkar is not greater than sagar
```

```python
#Example 2
omkar=19
sagar=19
print(omkar != sagar) #This statement shows False, because of operator tells omkar is not eaqual to sagar, but value of omkar and sagar are both equal.
print(omkar == sagar) #This statement shows True, because of operator tells omkar is equal to equal sagar and actua value of variables are equal.
```

# Conditional statment
## if :
### We use if for checking half condition only
## if-else :
### if-else statment used to check one complete condition
### syntax : if condition :
###                if block code
###         else :
###               else block code


```python
#Example of conditinal statement.
#program to check who will pay the bill 
sandesh=24
sagar=21
if (sandesh > sagar) :
    print("sandesh will pay the bill")
else :
    print("sagar will pay the bill")
```

```python
#Example1
#program using elif condition
sandesh=23
sagar=23
if (sandesh > sagar) :
    print("sandesh will pay the bill")
elif (sandesh == sagar) :
    print("TTMM")
else:
    print("sandesh will pay the bill")
print("End of program")
```

```python
#Example2
omkar=21
sagar=22
sandesh=23
if (omkar > sagar):
    print("omkar will pay the bill")
elif (sagar > sandesh) :
    print("sagar will pay the bill")
else :
    print("sandesh will pay the bill")
```
