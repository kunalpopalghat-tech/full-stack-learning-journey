# Q.What is tuple ?
## --> Tuple is im-mutable, Heterogenius collection of element where order is maintained and duplicates are allowed.

# Q.How to create tuple ?
syntax :---> t = (1,2,3,4) tuple is qauma seperated values inside parentheses. 

```python
t = (10,20,30,40)
print(type(t))
print(dir(t))
print(t.count(10)) # It shows total count of that element.if element is not fount it raise value error.
print(t.index(40)) # It shows index of given element. if element is not fount it raise value error.
```

## Create a tuple with single element ?

```python
t1 = (10)
print(type(t1))
t1 = (10,) # only quama seperated value is tuple .
print(type(t1))
```

# Q. How to access data from tuple ?
## --> Indexing and slicing 

```python
category = ("Electronics","Furniture","Electronics","Clothing")
print(category)
print(type(category))
# If you want to make changes in tuple. you have to convert tuple into list then make changes and again convert that list into tuple.
cat_list = list(category)
print(cat_list)
print(type(cat_list))
cat_list.append("stationary")
print(cat_list)
category = tuple(cat_list) # list converting into tuple after changes .
print(category)
```

## Packing and unpacking of elements / data .

```python
bag1 = ("clothes","charger","cash","whatever")
bag2 = ["clothes","charger","cash","whatever"]
print(type(bag1))
print(type(bag2)) 

print(bag1)
v1 , v2 , v3 , v4 = bag1 # unpacking 
print(v1 , v2 , v3 , v4 )

b1,b2,b3,b4 = bag2
print(b1,b2,b3,b4)
```

```python
t1 = 10
t2 = "name"
t3 = 10.5
t4 = "xyz"

bag1 = t1,t2,t3,t4 # Packing of elements 
print(bag1)
print(type(bag1))
```
