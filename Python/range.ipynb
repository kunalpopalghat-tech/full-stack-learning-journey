# Q. What is range in python ?
## --> Range is data type in python . It is used to create sequence of integer numbers -->1,2,3,4

# Q. How to create a range ?

### We can create a range by following ways :
#### 1. range(End)
#### 2. range(start,end)
#### 3. range(start,end,step)


##### 1.range(End)

```python
n = 12
r = range(n)
print(r)
print(type(r))
## Eterating elements :
for i in r:
    print(i)
```

##### 2.range(Start,End)


```python
r = range(3 , 30)
for num in r:
    print(num)
```

##### 3.range(Start,End,step)


```python
for num in range(5,51,5):
    print(num)
```
