# Python List methods 

### 1. append() :
The append() method is used to add one new element at the end of an existing list. It modifies the original list directly and does not create a new list. The element can be of any data type, such as a number, string, tuple, or even another list.

```python
## Example 
numbers = [10, 20, 30]
numbers.append(40)
print(numbers)
```

### 2. extend() :
The extend() method is used to add multiple elements to the end of an existing list. It takes an iterable such as a list, tuple, set, or string and adds its elements individually to the existing list. Unlike append(), it does not add the entire iterable as a single element.

```python
## Example 
numbers = [10, 20]
numbers.extend([30, 40, 50])
print(numbers)
```

### 3. insert() :
The insert() method is used to add an element at a specific index position in a list. It takes two arguments: the first argument specifies the index where the element should be inserted, and the second argument specifies the element that needs to be inserted. Existing elements are shifted to the right.

```python
## Example
numbers = [10, 20, 30]
numbers.insert(1, 15)
print(numbers)
```

### 4. remove() :
The remove() method is used to remove a specific value from a list. If the same value appears multiple times, it removes only the first occurrence of that value. If the specified value does not exist in the list, Python raises a ValueError.

```python
## Example
numbers = [10, 20, 30, 20]
numbers.remove(20)
print(numbers)
```

### 5. pop() :
The pop() method is used to remove and return an element from a list. If we do not provide an index, it removes and returns the last element of the list. We can also provide an index to remove an element from a specific position.

```python
## Example
numbers = [10, 20, 30]

value = numbers.pop()

print(value)
print(numbers)
```

### 6. clear() :
The clear() method is used to remove all elements from an existing list. After using this method, the list remains available but contains no elements, so it becomes an empty list. It modifies the original list directly.

```python
## Example

numbers = [10, 20, 30]

numbers.clear()

print(numbers)
```

### 7. index() :
The index() method is used to find the index position of a specified element in a list. If the same element occurs multiple times, it returns the index of its first occurrence. If the element is not present, Python raises a ValueError.

```python
## Example
numbers = [10, 20, 30, 20]

position = numbers.index(20)

print(position)
```

### 8. count() :
The count() method is used to determine how many times a particular element occurs in a list. It checks the complete list and returns an integer representing the total number of occurrences of that element. If the element is not present, it returns 0.

```python
## Example
numbers = [10, 20, 20, 30, 20]

result = numbers.count(20)

print(result)
```

### 9. sort() :
The sort() method is used to arrange the elements of a list in a particular order. By default, it sorts the elements in ascending order. We can use reverse=True to sort the elements in descending order. This method modifies the original list directly.

```python
## Example
numbers = [30, 10, 20]

numbers.sort()

print(numbers)

## for descending order

numbers.sort(reverse=True)

print(numbers)
```

### 10. reverse() :
The reverse() method is used to reverse the current order of elements in a list. It does not sort the elements; instead, it simply changes the order so that the first element becomes the last and the last element becomes the first. It modifies the original list directly.

```python
## Example
numbers = [10, 20, 30, 40]

numbers.reverse()

print(numbers)
```

### 11. copy() :
The copy() method is used to create a shallow copy of an existing list. It creates a new list containing the same elements as the original list, so changes made to the new list do not affect the original list for the copied list structure.

```python
## Example
numbers = [10, 20, 30]

new_numbers = numbers.copy()

print(new_numbers)
```
