### slicing with step value
 value of step value is by default is +1.

```python
# Extract every second character
word = "Python"
print(word[::2])
```

```python
# Extract every second character
word = "Programming"
print(word[::2])
```

```python
# Reverse the string using slicing
word = "Developer"
print(word[ : : -1])
```

```python
# Reverse the complete string
word = "Python Programming"
print(word[ : : -1])
```

```python
# Extract characters using start, stop and step
# Try to get: ACEGI
word = "ABCDEFGHIJ"
print(word[::2])
```


## Important Rules

1. Indexing starts from 0.
2. Negative indexing starts from -1.
3. Slicing syntax: [start:stop:step]
4. Stop index is excluded.
5. [::2] means every 2nd character.
6. [::-1] reverses the string.
